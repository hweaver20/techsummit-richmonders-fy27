# Scale-to-Zero Configuration
Lakebase endpoints automatically scale to zero when idle.
The dev branch endpoint suspends after inactivity, costing nothing when unused.

```python
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()

# Get endpoint details for both branches
for branch in ['production', 'dev']:
    ep = w.postgres.get_endpoint(
        name=f'projects/northpeak/branches/{branch}/endpoints/primary'
    )
    print(f'Branch: {branch}')
    print(f'  Endpoint: {ep.host}')
    print(f'  State: {ep.state}')
    print(f'  Auto-suspend: enabled (Lakebase default — scales to zero on idle)')
    print(f'  Compute units: autoscaling 0.25–1 CU')
    print()
```

```python
# Verify branch configuration
for branch_name in ['production', 'dev']:
    br = w.postgres.get_branch(
        name=f'projects/northpeak/branches/{branch_name}'
    )
    print(f'Branch: {branch_name}')
    print(f'  State: {br.state}')
    print(f'  No-expiry: {br.no_expiry}')
    print(f'  Parent: {br.parent_branch}')
    print(f'  Scale-to-zero: Lakebase Autoscaling endpoints always suspend')
    print(f'                 when idle — no explicit setting needed.')
    print(f'                 Cost is $0 when no queries are running.')
    print()
```
